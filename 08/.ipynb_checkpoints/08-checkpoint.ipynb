## <span style="color:blue">**Esercizio 08.1**

```python
import matplotlib
import matplotlib.pyplot as plt
import numpy as np
import math

fig, axes = plt.subplots(4, 3, figsize=(15, 15))  

for i in range(0, 4):
    data_energy = np.loadtxt(f"08.1/energy_{i}.dat", comments="#")
    data_accept = np.loadtxt(f"08.1/acceptance_{i}.dat", comments="#")
    data_amp = np.loadtxt(f"08.1/amplitude_{i}.dat", comments="#")

    blocchi = data_energy[:,0]
    energy = data_energy[:,2]
    error_energy = data_energy[:,3]
    
    acceptance = data_accept[:,1]

    x_amp = data_amp[:,0]
    amp = data_amp[:,1]

    axes[i,0].errorbar(blocchi, energy , yerr=error_energy, color='b')
    axes[i,0].set_title("prova")
    axes[i,0].set_xlabel("Numero di blocchi")
    axes[i,0].set_ylabel("prova")
    axes[i,0].legend()
    axes[i,0].grid(True)

    axes[i,1].plot(blocchi, acceptance , color='r')
    axes[i,1].set_xlabel("Numero di blocchi")
    axes[i,1].set_ylabel("prova")
    axes[i,1].grid(True)

    axes[i,2].plot(x_amp, amp , color='g')
    axes[i,2].set_xlabel("x")
    axes[i,2].set_ylabel("$|\psi|^2$")
    axes[i,2].grid(True)


plt.show()
    
```
