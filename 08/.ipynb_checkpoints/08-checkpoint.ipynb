## <span style="color:blue">**Esercizio 08.1**

```python
import matplotlib.pyplot as plt
import numpy as np
import seaborn as sns

# Impostazioni estetiche
sns.set_theme(style="whitegrid", palette="muted")
plt.rcParams.update({
    "figure.figsize": (15, 15),
    "axes.titlesize": 14,
    "axes.labelsize": 12,
    "xtick.labelsize": 10,
    "ytick.labelsize": 10,
    "legend.fontsize": 10,
    "font.family": "serif"
})

fig, axes = plt.subplots(4, 3, figsize=(15, 15), constrained_layout=True)

for i in range(0,4):
    data_energy = np.loadtxt(f"08.1/energy_{i}.dat", comments="#")
    data_accept = np.loadtxt(f"08.1/acceptance_{i}.dat", comments="#")
    data_amp = np.loadtxt(f"08.1/amplitude_{i}.dat", comments="#")

    blocchi = data_energy[:, 0]
    energy = data_energy[:, 2]
    error_energy = data_energy[:, 3]
    acceptance = data_accept[:, 1]
    x_amp = data_amp[:, 0]

    # Energia media
    axes[i, 0].errorbar(blocchi, energy, yerr=error_energy, color='royalblue', label="Energia media")
    axes[i, 0].set_title(f"Configurazione {i} - Energia")
    axes[i, 0].set_xlabel("Numero di blocchi")
    axes[i, 0].set_ylabel(r"$\langle E \rangle$")
    axes[i, 0].legend()
    axes[i, 0].grid(True)

    # Accettanza
    axes[i, 1].plot(blocchi, acceptance, color='crimson')
    axes[i, 1].set_title(f"Configurazione {i} - Accettanza")
    axes[i, 1].set_xlabel("Numero di blocchi")
    axes[i, 1].set_ylabel("Accettanza")
    axes[i, 1].grid(True)

    # Istogramma densità di probabilità
    axes[i, 2].hist(x_amp, bins=100, density=True, color='mediumseagreen', alpha=0.7)
    axes[i, 2].set_title(f"Configurazione {i} - Densità campionata")
    axes[i, 2].set_xlabel("x")
    axes[i, 2].set_ylabel(r"$|\psi(x)|^2$")
    axes[i, 2].grid(True)

# Mostra il grafico
plt.suptitle("Analisi delle simulazioni per diverse configurazioni $(\mu, \sigma)$", fontsize=16)
plt.show()
```

## <span style="color:blue">**Esercizio 08.2**

```python
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D

T, mu, sigma = np.loadtxt("08.2/parameters.dat", unpack=True, usecols=(0,1,2), skiprows=1)
beta = 1/T

fig = plt.figure(figsize=(10, 8))
ax = fig.add_subplot(111, projection='3d')

#Scatter 3D con gradiente di colore in base a beta,
sc = ax.scatter(beta, mu, sigma,
                c=beta,              # colore in base a beta
                cmap='viridis',      # colormap (puoi provare anche 'plasma', 'inferno', ecc.)
                marker='.',          # marker leggero
                s=20,                # dimensione marker
                alpha=0.8,           # opacità
                label='mu-sigma vs beta')

#Etichette,
ax.set_xlabel('Beta', fontsize=14)
ax.set_ylabel('Mu', fontsize=14)
ax.set_zlabel('Sigma', fontsize=14)
ax.set_title('3D plot con gradiente di colore su Beta', fontsize=16)
ax.tick_params(axis='both', which='major', labelsize=12)

#Barra del colore,
cbar = fig.colorbar(sc, ax=ax, pad=0.1)
cbar.set_label('Beta', fontsize=12)

#Legenda,
ax.legend(fontsize=12)

plt.tight_layout()
plt.show()
```
