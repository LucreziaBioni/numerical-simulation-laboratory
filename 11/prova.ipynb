```python
import os
os.environ["TF_CPP_MIN_LOG_LEVEL"] = "3"

import tensorflow as tf
import numpy as np
import matplotlib.pyplot as plt

# Generiamo dati fittizi: y = 3x + 2 + rumore
np.random.seed(0)
x = np.random.rand(100).astype(np.float32)
y = 3 * x + 2 + np.random.normal(0, 0.1, size=x.shape).astype(np.float32)

# Costruiamo un modello sequenziale semplice
model = tf.keras.Sequential([
    tf.keras.layers.Dense(units=1, input_shape=[1])
])

# Compiliamo il modello con loss e ottimizzatore
model.compile(optimizer='sgd', loss='mean_squared_error')

# Alleniamo il modello
history = model.fit(x, y, epochs=100, verbose=0)

# Stampiamo il coefficiente appreso
weights = model.get_weights()
print(f"Peso appreso: {weights[0][0][0]:.2f}, Bias appreso: {weights[1][0]:.2f}")

# Confronto visivo: dati reali vs predetti
y_pred = model.predict(x)

plt.scatter(x, y, label='Dati reali')
plt.plot(x, y_pred, color='red', label='Predizione')
plt.legend()
plt.title("TensorFlow: Regressione Lineare")
plt.xlabel("x")
plt.ylabel("y")
plt.show()
```
