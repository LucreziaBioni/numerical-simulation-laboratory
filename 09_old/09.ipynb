## <span style="color:blue">**Esercizio 09.1**

```python
import matplotlib.pyplot as plt
import numpy as np

name_cities, x, y = np.loadtxt("cities.dat", unpack=True, usecols=(0,1,2), dtype=str)
x = x.astype(float)
y = y.astype(float)

best_route = np.loadtxt("best_route.dat", dtype=str)


n_cities = len(best_route)
best_route_coord = np.zeros((2, n_cities))  # 2 righe: x e y

for i in range(n_cities):
    city_name = best_route[i]
    index = np.where(name_cities == city_name)[0][0]  # trova indice nella lista originale

    best_route_coord[0, i] = x[index]
    best_route_coord[1, i] = y[index]

plt.plot(best_route_coord[0], best_route_coord[1], 'o-', label='Best route')
plt.plot(x, y, 'o', label='Cities')
theta = np.linspace(0, 2*np.pi, 300)
plt.plot(np.cos(theta), np.sin(theta))
plt.gca().set_aspect('equal')
plt.xlabel("x")
plt.ylabel("y")
plt.legend()
plt.grid()
plt.show()
```
