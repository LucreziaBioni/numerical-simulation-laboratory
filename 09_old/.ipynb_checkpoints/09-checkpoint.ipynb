```python
def leggi_citta(file_path):
    città_coord = {}
    with open(file_path, 'r') as f:
        for riga in f:
            parti = riga.strip().split()
            if len(parti) != 3:
                continue
            nome, x, y = parti
            città_coord[nome] = (float(x), float(y))
    return città_coord

def leggi_best_route(file_path):
    with open(file_path, 'r') as f:
        return [riga.strip() for riga in f if riga.strip()]

def costruisci_vettore_coord(route, città_coord):
    return [città_coord[nome] for nome in route if nome in città_coord]

# File input
file_citta = 'cities.dat'
file_route = 'best_route'

# Caricamento dati
città_coord = leggi_citta(file_citta)
route = leggi_best_route(file_route)
vettore_coord = costruisci_vettore_coord(route, città_coord)

# Output di controllo
print("Coordinate nel miglior percorso:")
for coord in vettore_coord:
    print(coord)
```
